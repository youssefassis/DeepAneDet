```python
train, valid, test = dio.readSplit("/path/to/0Work/BaseConfig/CHUV/fold1.json")
patients = sorted(train + valid + test)
```

```python

for patient in patients:
    os.chdir(patient)
    # load config.json file
    with open('config.json') as f:
        config = json.load(f)
    # load volume
    vol, v2m = dio.read_nii_from_file(config['init volume'])
    
    if os.path.isfile("F.csv"):
        if "healthy" in patient:
            print(patient)
        spheres = dio.points_to_spheres(dio.read_points_from_csv("F.csv"))
        for i, s in enumerate(spheres):
            t = np.zeros(vol.shape).astype(np.uint8)
            ved.drawSphere(t, v2m, s[:3], s[3], 1)
            print(os.path.join(patient, "_".join(patient.split("/")[-2:])+f"_desc-Lesion_{i+1}_mask.nii.gz"))
#            break
            dio.save_nii_to_file("_".join(patient.split("/")[-2:])+f"_desc-Lesion_{i+1}_mask.nii.gz", t, v2m)


```
